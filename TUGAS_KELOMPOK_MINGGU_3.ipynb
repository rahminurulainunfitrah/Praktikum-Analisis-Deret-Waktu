## MA (1)

Proses MA(1) memiliki persamaan:

<center> $Y_t=e_t-\theta e_{t-1}$ </center>

#1. SIMULASI DERET MA(1)

Dengan panjang pengamatan 20 dengan nilai parameter:

### 1. MA(1) θ=0.1

```python
np.random.seed(113)
theta=0.1
et=[np.random.normal(0,1)]
ma1_a=[]
for i in range(20):
    et.append(np.random.normal(0,1))
    ma1_a.append(et[i+1]-theta*et[i])
```

```python
plt.figure()
plt.plot(ma1_a)
plt.xlabel('Waktu'), plt.ylabel(r'$Y_t$')
plt.show()
```

```python
#lag-1
plt.figure(figsize=(6,5))
pd.plotting.lag_plot(pd.DataFrame(ma1_a), lag=1, c='k')
plt.xlabel(r'$Y_{t-1}$'),plt.ylabel(r'$Y_{t}$')
plt.show()
```

```python
plt.figure(figsize=(6,5))
pd.plotting.lag_plot(pd.DataFrame(ma1_a), lag=2, c='k')
plt.xlabel(r'$Y_{t-2}$'),plt.ylabel(r'$Y_{t}$')
plt.show()
```

```python
plt.figure(figsize=(6,5))
pd.plotting.lag_plot(pd.DataFrame(ma2_a), lag=2, c='k')
plt.xlabel(r'$Y_{t-2}$'),plt.ylabel(r'$Y_{t}$')
plt.show()
```

```python
plt.figure(figsize=(6,5))
pd.plotting.lag_plot(pd.DataFrame(ma2_a), lag=3, c='k')
plt.xlabel(r'$Y_{t-3}$'),plt.ylabel(r'$Y_{t}$')
plt.show()
```

# **AR (1)**
Proses AR (1) memiliki persamaan:
$Y_t = \phi Y_{t-1} + e_t$

Syarat kestasioneran untuk deret AR (1) adalah jika dan hanya jika
$|\phi| < 1$


Untuk $\phi=0.3$

```python
np.random.seed(113)
AR1_1 = np.array([1,-0.3])
MA1_1 = np.array([1])
ar_object_1 = ArmaProcess(AR1_1,MA1_1)
AR1_1 = ar_object_1.generate_sample(nsample=100)
```

# **AR (2)**

<h2> AR (2) </h2>

Proses AR (2) memiliki persamaan:

<center> $Y_t=\phi_1 Y_{t-1}+\phi_2 Y_{t-2}+e_t$ </center>

**Syarat kestasioneran untuk deret AR (2) adalah jika dan hanya jika tiga kondisi berikut terpenuhi, yaitu:**

**<center> $\phi_1+\phi_2<1$  ,   $\phi_2-\phi_1<1$ ,   $|\phi_2|<1$ </center>**

Untuk $\phi_1=1.5$ dan $\phi_2=-0.5$

```python
np.random.seed(113)
phi1_AR2_1, phi2_AR2_1 = 1.5, -0.5

ar_params_1 = np.array([1, -phi1_AR2_1, -phi2_AR2_1]) #nilai parameter theta dalam sintaks dituliskan sebagai negatif dari nilai parameter asli
ma_params_1 = np.array([1])

AR_object2_1 = ArmaProcess(ar_params_1, ma_params_1)
AR2_s_1 = AR_object2_1.generate_sample(nsample=100)

plt.figure(figsize=(10,5))
plt.plot(AR2_s_1,'-o')
plt.xlabel('Waktu'), plt.ylabel(r'$Y_t$')
plt.title(f'Simulasi Data AR(2)')
plt.show()
```

```python
plot_acf(np.array(AR2_s_1), zero=False)
plt.show()
```
